### Concepts in plain language

A list is an ordered sequence and can contain duplicates. A tuple is a fixed sequence whose entries cannot be reassigned. A dictionary maps unique keys to values. A set keeps unique values and is useful for membership tests. Choose based on the question: preserve order, count occurrences, or find uniqueness?

A loop visits one item at a time. An accumulator stores the partial result. `counts.get(word, 0)` means use the current count if present, otherwise start from zero. A copy is a separate container; assigning `other = original` simply gives the same container another name.

Normalization makes a deliberate comparison policy: stripping spaces and lowercasing can make email identifiers consistent. Text tokenization means deciding where words begin and end. The course's beginner analyzer uses ASCII letters; it does not cover every language.

## Lesson

A list preserves order and allows changes. A tuple is useful for a fixed record or immutable key. A set stores unique values and supports membership and set operations. A dictionary maps unique keys to values; it is useful for counting and indexing.

A loop processes one element at a time. An accumulator collects partial results. A comprehension builds a collection from a simple expression; use a normal loop when validation or several steps would make a comprehension hard to read. Normalize data once at a boundary. For text counting, lowercase and punctuation rules must be explicit because “AI” and “ai” may otherwise become separate words. Do not modify a collection while iterating over it unless the behavior is deliberate.

## Exercise 1: Word frequencies

Count lowercase words matching [a-z]+; return a dictionary. Empty text returns {}.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
import re
text = 'Ai, ai!3da ta 2Data a32'
re.findall(r"[a-z]+", text.lower())
# for word in re.findall(r"[a-z]+", text.lower()):
````

````python
def word_counts(text):
    import re
    counts = {}
    for word in re.findall(r'[a-z]+',text.lower()):
        counts[word] = counts.get(word,0) + 1;
    return counts
````

````python
assert word_counts("AI, ai! Data.") == {"ai": 2, "data": 1}
assert word_counts("") == {}
print("Exercise 1: checks passed")
````

## Exercise 2: Stable deduplication

Return the first occurrence of each hashable item, preserving input order.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def unique(items):
    seen, results = set() , [] 
    for item in items:
        if item not in seen:
            seen.add(item)
            results.append(item)
    return results
````

````python
assert unique([3, 1, 3, 2, 1]) == [3, 1, 2]
assert unique([]) == []
print("Exercise 2: checks passed")
````

## Exercise 3: Inventory update

Return a new dictionary with deltas applied. Reject a resulting negative count. Do not change the original.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def update_stock(stock, deltas):
    updatedStock = dict(stock)
    for item, change in deltas.items():
        after_change = updatedStock.get(item, 0) + change;
        if(after_change < 0):
            raise ValueError('Stock cannot be less than 0')
        updatedStock[item] = after_change
    return updatedStock
````

````python
original = {"pen": 3}
assert update_stock(original, {"pen": -2, "book": 1}) == {"pen": 1, "book": 1}
assert original == {"pen": 3}
expect_error(ValueError, lambda: update_stock(original, {"pen": -4}))
print("Exercise 3: checks passed")
````
