### 1. From a number to a function

A **scalar** is one number, such as a price of 4. A **variable** is a name for a number that may change. In `cost = 4 * quantity`, changing `quantity` changes `cost`. A mathematical function writes this as $f(q)=4q$. Here $f$ names the rule, $q$ is the input, and $f(q)$ is the output; parentheses do not mean multiplication in this notation.

An **equation** says two expressions have equal values. Solving $3x+2=14$ means finding the value of $x$: subtract 2 from both sides to get $3x=12$, then divide both sides by 3 to get $x=4$. Check by substituting: $3(4)+2=14$. In Python, `=` assigns a value and `==` compares values; neither automatically solves an equation.

A line $y=mx+b$ has **slope** $m$ (change in $y$ per one-unit increase in $x$) and **intercept** $b$ (the value at $x=0$). For $y=3x+2$, inputs 0, 1, 2 produce outputs 2, 5, 8. Plot points $(x,y)$: move horizontally by $x$, then vertically by $y$. The line connects these points. A graph is a picture of the input/output rule, not a different kind of calculation.

### 2. Powers, roots, and logarithms

$x^2=x\times x$, so $(-3)^2=9$. The nonnegative square root $\sqrt{9}=3$ asks which nonnegative number squares to 9. Python uses `x ** 2`; `x ^ 2` is a different operation called bitwise XOR.

A logarithm reverses an exponent: $2^3=8$ means $\log_2(8)=3$. The natural logarithm $\ln(x)$ uses base $e\approx2.718$; `math.log(x)` computes it. Real logarithms here require $x>0$. Logarithms turn multiplication into addition: $\ln(ab)=\ln(a)+\ln(b)$ for positive inputs. Later, this makes products of small probabilities easier to handle.

### 3. A vector is an ordered collection

The vector $v=[3,4]$ can describe a displacement: three units right and four units up. It could instead describe two features of a record; always state what each position means. Order matters: `[3,4]` and `[4,3]` describe different vectors. Its **dimension** is the number of entries, here 2.

Vector addition combines matching entries: $[1,2]+[3,5]=[4,7]$. Scaling multiplies each entry: $2[1,2]=[2,4]$. Ordinary Python lists behave differently: `[1,2] + [3,5]` concatenates, and `2 * [1,2]` repeats the list. Implement entry-by-entry arithmetic or use arrays when introduced later.

The **dot product** multiplies matching entries and adds: $[2,1]\cdot[3,4]=2(3)+1(4)=10$. Two equal-length vectors produce one scalar. A weighted bill is a dot product: quantities times unit prices, summed. Check units before interpreting the result.

The **Euclidean norm** is length: $\|[3,4]\|=\sqrt{3^2+4^2}=5$. The double vertical bars mean norm, not two divisions. Distance between $u$ and $v$ is $\|u-v\|$. Between $[1,1]$ and $[4,5]$, subtract first to get $[3,4]$, then take length 5.

### 4. Direction and cosine similarity

Cosine similarity is $(u\cdot v)/(\|u\|\|v\|)$. The numerator measures alignment; the denominator removes the effect of lengths. Same direction gives 1, perpendicular directions give 0, opposite directions give -1. For `[1,0]` and `[0,1]`, the dot product is 0, so similarity is 0. A zero vector has no direction and makes the denominator zero: report an error or an explicit undefined result, not an invented similarity.

### 5. Matrices and shapes

A **matrix** is a rectangular table. `[[1,2,3],[4,5,6]]` has 2 rows and 3 columns, written shape $2\times3$. The **transpose** swaps rows and columns, producing `[[1,4],[2,5],[3,6]]`, shape $3\times2$. This is not sorting or reversing the entries.

Matrix multiplication takes a row from the first matrix and a column from the second, then computes their dot product. Shapes $(m\times n)(n\times p)$ produce $m\times p$: the inside dimensions must match. For `[[1,2],[3,4]]` times `[[5],[6]]`, the output is `[[1*5+2*6],[3*5+4*6]] = [[17],[39]]`. Multiplication is not entry-by-entry multiplication, and reversing the order usually changes the answer or makes the shapes invalid.

### 6. Transformations, angles, and eigenvectors

A matrix can describe a rule that transforms vectors. The identity matrix `[[1,0],[0,1]]` leaves every vector unchanged. `[[2,0],[0,3]]` stretches horizontal components by 2 and vertical components by 3.

Angles in Python's trigonometric functions are in **radians**: a full turn is $2\pi$, a half turn is $\pi$, and a quarter turn is $\pi/2$. On the unit circle, cosine is the horizontal coordinate and sine is the vertical coordinate. A counterclockwise rotation uses rows `[cos(theta), -sin(theta)]` and `[sin(theta), cos(theta)]`. At a quarter turn this becomes `[[0,-1],[1,0]]`; `[1,0]` turns into `[0,1]`, preserving length.

An **eigenvector** is a nonzero vector that a matrix only scales, rather than turning into a different line. $Av=\lambda v$ reads “matrix $A$ acting on vector $v$ equals scalar $\lambda$ times that vector.” For `A=[[4,0],[0,2]]`, `v=[1,0]` becomes `[4,0]`, so its eigenvalue is 4. The zero vector is excluded because it would satisfy that equation for every eigenvalue.

## Lesson

A scalar is one number, a vector an ordered collection, and a matrix a rectangular collection. A tensor generalizes dimensions. A dot product sums pairwise products and requires equal lengths. A matrix A of shape (m,n) can multiply B of shape (n,p), producing (m,p). Each output is a row-column dot product.

The Euclidean norm measures length. Cosine similarity divides a dot product by both vector norms, measuring direction rather than magnitude; it is undefined for a zero vector. A transpose swaps axes. An identity matrix leaves vectors unchanged. An eigenvector keeps its direction under a matrix transformation: Av = λv. Embeddings are learned vectors, so geometry helps explain retrieval.

Algebra expresses relationships with functions: y=mx+b, powers, logarithms, and summations. Logarithms turn products into sums and appear in likelihoods. Always check the domain: log requires positive real inputs.

## Exercise 1: Dot product

Return sum(a_i*b_i); reject unequal lengths.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def dot(a, b):
    if len(a) != len(b) : raise ValueError("Vector shape mismatch")
    return sum(f*s for f,s in zip(a,b))
````

````python
assert dot([1, 2], [3, 4]) == 11
expect_error(ValueError, lambda: dot([1], [2, 3]))
print("Exercise 1: checks passed")
````

## Exercise 2: Cosine similarity

Use dot to return cosine similarity; reject a zero vector.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def cosine(a, b):
    if(len(a) != len(b)): raise ValueError("Shape mismatch")
    magnitude_a = math.sqrt(sum(x*x for x in a))
    magnitude_b = math.sqrt(sum(x*x for x in b))
    product_of_magnitude = magnitude_a * magnitude_b 
    if(product_of_magnitude == 0): raise ValueError('Zero Vecotr')
    a_dot_b = sum(x*y for x,y in zip(a,b))
    return a_dot_b/product_of_magnitude
````

````python
assert abs(cosine([1, 0], [0, 1])) < 1e-12
assert abs(cosine([2, 2], [1, 1])-1) < 1e-12
expect_error(ValueError, lambda: cosine([0, 0], [1, 2]))
print("Exercise 2: checks passed")
````

## Exercise 3: Matrix multiplication

Multiply nonempty rectangular nested lists. Validate rectangularity and inner dimensions.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def matmul(a, b):
    if not a or not b or not a[0] or not b[0]: raise ValueError("Empty matrix")
    if any(len(row) != len(a[0]) for row in a) or any(len(row) != len(b[0]) for row in b):
        raise ValueError("Ragged matrix")
    if len(a[0]) != len(b): raise ValueError("Shape mismatch")
    result = []
    for row_a in a:
        new_row = []
        for col_index in range(len(b[0])):
            col_b = [row[col_index] for row in b]
            value = dot(row_a,col_b)
            new_row.append(value)
        result.append(new_row)
    return result
````

````python
assert matmul([[1, 2]], [[3], [4]]) == [[11]]
assert matmul([[2, 3]], [[1, 0], [0, 1]]) == [[2, 3]]
expect_error(ValueError, lambda: matmul([[1, 2]], [[1, 2]]))
print("Exercise 3: checks passed")
````
