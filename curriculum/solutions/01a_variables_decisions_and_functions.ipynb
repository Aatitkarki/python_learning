### Concepts in plain language

A string such as `"12"` is text; the integer `12` is a number. `input()` returns text, so validate and convert it before arithmetic. Booleans (`True`, `False`) are values used in conditions. `if` chooses a branch; `while` repeats while its condition remains true.

A function is a named operation. Parameters are the names in its definition; arguments are the values passed to it. `return` hands a result back to the caller. `print` only displays a result. Local variables exist inside a function call. Indentation tells Python which statements belong to the function or branch.

An exception signals that normal completion is impossible. Catch a specific expected error such as invalid numeric input at the user boundary. Catching every exception can hide a bug in your program.

## Exercise 1: Profit function

Return (sell-buy)*shares. Require shares > 0 and prices >= 0.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def profit(buy, sell, shares):
    if shares <= 0 or min(buy, sell) < 0:
        raise ValueError("Invalid price or quantity")
    return (sell - buy) * shares
````

````python
assert profit(100, 110, 10) == 100
assert profit(10, 8, 2) == -4
expect_error(ValueError, lambda: profit(1, 2, 0))
print("Exercise 1: checks passed")
````

**Why this works:** Separate invalid inputs from valid but undesirable business outcomes.


## Exercise 2: Calculator

Support +, -, *, /. Reject unknown operators with ValueError and division by zero with ZeroDivisionError.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def calculate(a, op, b):
    if op == "+": return a + b
    if op == "-": return a - b
    if op == "*": return a * b
    if op == "/": return a / b
    raise ValueError("Unknown operator")
````

````python
assert calculate(6, "/", 2) == 3
assert calculate(6, "-", 9) == -3
expect_error(ZeroDivisionError, lambda: calculate(2, "/", 0))
expect_error(ValueError, lambda: calculate(2, "**", 4))
print("Exercise 2: checks passed")
````

**Why this works:** An allowlist keeps the behavior small, readable, and testable.


## Exercise 3: Tiered shipping

Orders >= 100 ship free; smaller nonnegative orders cost 5. A negative total raises ValueError.

Write down one normal case, one boundary case, and one invalid case before coding.


````python
def shipping(total):
    if total < 0: raise ValueError("Negative total")
    return 0 if total >= 100 else 5
````

````python
assert [shipping(x) for x in (0, 99, 100)] == [5, 5, 0]
expect_error(ValueError, lambda: shipping(-1))
print("Exercise 3: checks passed")
````

**Why this works:** Boundary examples find off-by-one and incorrect comparison operators.
