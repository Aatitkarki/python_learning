## Exercise 1: Study budget

Return total hours for nonnegative weeks and hours_per_week; reject negative inputs.

Write down one normal case, one boundary case, and one invalid case before coding.


```python
def study_hours(weeks, hours_per_week):
    if weeks < 0 or hours_per_week < 0:
        raise ValueError("Time cannot be negative")
    return weeks * hours_per_week
```

```python
assert study_hours(80, 15) == 1200
assert study_hours(0, 15) == 0
expect_error(ValueError, lambda: study_hours(-1, 15))
print("Exercise 1: checks passed")
```

**Why this works:** The unit is weeks × hours/week = hours. Validation makes your contract explicit.


## Exercise 2: Experiment manifest

Return a dictionary containing seed and Python version (under keys seed, python). Import sys inside your function.

Write down one normal case, one boundary case, and one invalid case before coding.


```python
def manifest(seed):
    import sys
    return {"seed": seed, "python": sys.version.split()[0]}
```

```python
assert manifest(42)["seed"] == 42
assert len(manifest(42)["python"].split(".")) == 3
print("Exercise 2: checks passed")
```

**Why this works:** A seed alone is not full reproducibility: record dependencies, data hashes, hardware, and code revision too.


## Exercise 3: Reproducible randomness

Return n integer die rolls using a private random.Random(seed).

Write down one normal case, one boundary case, and one invalid case before coding.


```python
def rolls(n, seed):
    import random
    rng = random.Random(seed)
    return [rng.randint(1, 6) for _ in range(n)]
```

```python
assert rolls(20, 7) == rolls(20, 7)
assert len(rolls(20, 7)) == 20
assert all(1 <= x <= 6 for x in rolls(20, 7))
print("Exercise 3: checks passed")
```

**Why this works:** A private generator prevents unrelated code from changing your experiment.
